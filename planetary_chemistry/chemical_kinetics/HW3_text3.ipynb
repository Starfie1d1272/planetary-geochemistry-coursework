```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import linregress

# 1. 定义理想气体常数
R = 8.3144

# 2. 定义已知数据序列 (转换为 numpy array 以便后续进行向量化计算)
T_celsius = np.array([1160, 1120, 1075, 1070, 1050])
Ni_Ol = np.array([1555, 1310, 955, 935, 840])
Ni_Cpx = np.array([255, 245, 240, 235, 220])

print("依赖包导入成功，基础数据已加载。")
```

```python
# 摄氏度转换为开尔文温度
T_kelvin = T_celsius + 273.15

# 计算分配系数 Kd = Ni_Ol / Ni_Cpx
Kd = Ni_Ol / Ni_Cpx

# 构建汇总表格 DataFrame
df_data = pd.DataFrame({
    "温度(°C)": T_celsius,
    "温度(K)": T_kelvin,
    "Ni_Ol (ppm)": Ni_Ol,
    "Ni_Cpx (ppm)": Ni_Cpx,
    "Kd": Kd
})

print("计算完成，数据汇总表如下：")
display(df_data) if 'display' in globals() else print(df_data)
```

```python
# 计算自变量 x 和因变量 y
x = 1 / T_kelvin
y = np.log(Kd) # np.log 默认以 e 为底，即自然对数 ln

# 使用 scipy 的 linregress 执行线性回归拟合
slope, intercept, r_value, p_value, std_err = linregress(x, y)
a = slope
b = intercept

# 计算热力学参数
delta_H = -a * R
delta_S = b * R

# 打印结果并格式化输出温度计方程
print(f"【拟合参数】")
print(f"斜率 a = {a:.2f}")
print(f"截距 b = {b:.4f}")
print(f"相关系数 R^2 = {r_value**2:.4f}\n")

print(f"【热力学参数】")
print(f"焓变 ΔH = {delta_H:.2f} J/mol")
print(f"熵变 ΔS = {delta_S:.4f} J/(mol·K)\n")

print(f"【温度计方程】")
print(f"T(K) = {a:.2f} / (ln(Kd) - ({b:.4f}))")
```

```python
# 创建绘图画布
plt.figure(figsize=(8, 6))

# 绘制实验数据的散点图
plt.scatter(x, y, color='blue', marker='o', s=80, label='Experimental Data', zorder=5)

# 绘制拟合直线
# 生成一条用于绘制平滑直线的 x 数据
x_line = np.linspace(min(x) * 0.98, max(x) * 1.02, 100)
y_line = a * x_line + b
plt.plot(x_line, y_line, color='red', linestyle='--', linewidth=2, label='Fitted Line')

# 绘图设置
plt.xlabel('1/T (K$^{-1}$)', fontsize=12)
plt.ylabel('ln(Kd)', fontsize=12)
plt.title('ln(Kd) vs 1/T for Ni in Ol-Cpx', fontsize=14)

# 将温度计方程作为文本添加到图表中的合适位置
# 这里我们使用坐标轴的相对坐标 (0.05, 0.90) 放置文本，防止遮挡数据点
eq_text = f"Equation: $y = {a:.0f}x + ({b:.2f})$\nThermometer: $T(K) = {a:.0f} / (\ln(K_D) - {b:.2f})$"
plt.text(0.05, 0.85, eq_text, transform=plt.gca().transAxes, 
         fontsize=12, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

# 开启网格线并添加图例
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(loc='lower right', fontsize=11)

# 调整布局并显示图形
plt.tight_layout()
plt.show()
```
