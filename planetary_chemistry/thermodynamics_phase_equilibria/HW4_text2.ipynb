```python
import numpy as np
import math

# Global physical constants
R_gas = 8.314  # J/(mol*K)
g = 9.81       # m/s^2

# Temperatures
T_cf = 300     # K, centrifuge
T_atm = 288    # K, atmosphere
```

```python
# IR-6 centrifuge parameters
r = 0.1      # m
f = 565      # Hz
omega = 2 * math.pi * f

delta_M_U = 0.003  # kg/mol, M(238UF6) - M(235UF6)

# Radial basic separation factor
q_0 = np.exp((delta_M_U * omega**2 * r**2) / (2 * R_gas * T_cf))

print(f"角速度 omega = {omega:.5f} rad/s")
print(f"径向 235U/238U 基本分离系数 q_0 = {q_0:.5f}")
```

```python
# Troposphere height difference
delta_z = 10000  # m

# Molar mass differences (kg/mol)
delta_M_N2 = 0.001
delta_M_O2_N2 = 0.004

# Isotopic ratio change for 15N14N / 14N14N from sea level to 10 km
ratio_N2_iso = np.exp(-(delta_M_N2 * g * delta_z) / (R_gas * T_atm))

# Molecular ratio change for N2 / O2 from sea level to 10 km
# N2 is lighter, so N2/O2 increases with altitude
ratio_N2_O2 = np.exp((delta_M_O2_N2 * g * delta_z) / (R_gas * T_atm))

change_N2_iso_pct = (ratio_N2_iso - 1.0) * 100
change_N2_O2_pct = (ratio_N2_O2 - 1.0) * 100

print(f"10 km 处氮同位素比值变化率 R(z)/R(z0) = {ratio_N2_iso:.6f}")
print(f"对应变化 = {change_N2_iso_pct:.3f}%")
print(f"10 km 处 N2/O2 比值变化率 = {ratio_N2_O2:.6f}")
print(f"对应变化 = +{change_N2_O2_pct:.3f}%")
```

### 为什么地球大气在对流层并未出现明显重力分层？

按玻尔兹曼分布，若只考虑静止流体中的分子扩散，重分子会相对富集在低处、轻分子在高处略增。

但真实对流层中，这种“微观扩散分层”通常不会显著出现，核心原因是：

1. **热力对流（Macroscopic convection）很强**：地表受太阳加热后，暖空气上升、冷空气下沉，形成持续的垂直质量交换。
2. **湍流混合（Turbulent mixing）很强**：天气系统、风切变与边界层过程引发湍流，使气体组分快速均匀化。
3. **混合时间尺度远短于分子扩散分离时间尺度**：宏观机械混合速度远大于重力场下缓慢的分子扩散分层。

因此，在对流层（约 0-10 km）内，大气通常可视为“充分混合层”，主要气体体积分数（如 N2、O2）随高度变化很小，不会形成显著的重力同位素或分子分层。
