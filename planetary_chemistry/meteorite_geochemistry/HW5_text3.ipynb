```python
# Organize data by class
from collections import defaultdict

groups = defaultdict(lambda: {"Co": [], "Ga": [], "Au": []})
for name, co, ga, au, cls in data:
    groups[cls]["Co"].append(co)
    groups[cls]["Ga"].append(ga)
    groups[cls]["Au"].append(au)

# Define plot style for each class
class_style = {
    "IAB-MG":   {"marker": "o", "color": "#E63946", "label": "IAB-MG",   "size": 80},
    "IAB-sLH":  {"marker": "^", "color": "#F4A261", "label": "IAB-sLH",  "size": 80},
    "IAB-sLL":  {"marker": "v", "color": "#E9C46A", "label": "IAB-sLL",  "size": 80},
    "IAB-sLM":  {"marker": "<", "color": "#2A9D8F", "label": "IAB-sLM",  "size": 80},
    "IAB-sHH":  {"marker": "P", "color": "#264653", "label": "IAB-sHH",  "size": 100},
    "IAB-sHL":  {"marker": "p", "color": "#7B2D8B", "label": "IAB-sHL",  "size": 100},
    "IAB-ungr": {"marker": "X", "color": "#D4A5A5", "label": "IAB-ungr", "size": 70},
    "IC":       {"marker": "D", "color": "#6D597A", "label": "IC",       "size": 70},
    "IIAB":     {"marker": "s", "color": "#1D3557", "label": "IIAB",     "size": 80},
    "IID":      {"marker": "d", "color": "#457B9D", "label": "IID",      "size": 90},
    "IIE":      {"marker": "H", "color": "#B5838D", "label": "IIE",      "size": 80},
    "IIIAB":    {"marker": ".", "color": "#2A9D8F", "label": "IIIAB",    "size": 120},
    "IIIE":     {"marker": "+", "color": "#E76F51", "label": "IIIE",     "size": 120},
    "IVA":      {"marker": "*", "color": "#9C89B8", "label": "IVA",      "size": 140},
    "Mes":      {"marker": "h", "color": "#F08080", "label": "Mes",      "size": 80},
    "ungr":     {"marker": "x", "color": "#999999", "label": "ungr",     "size": 60},
}

print("Chemical groups and sample counts:")
for cls in sorted(groups.keys()):
    n = len(groups[cls]["Co"])
    print(f"  {cls:12s}: {n} samples")
```

```python
# ========== 图5: Co vs Au 分布图（去除极端ungr样品） ==========
fig, ax = plt.subplots(1, 1, figsize=(11, 7))

for cls, sty in class_style.items():
    if cls not in groups or cls == 'ungr':
        continue
    ax.scatter(groups[cls]["Au"], groups[cls]["Co"],
              marker=sty["marker"], c=sty["color"],
              s=sty["size"], edgecolors='black', linewidth=0.5,
              label=sty["label"], zorder=3)

ax.set_xlabel('Au (μg/g)')
ax.set_ylabel('Co (mg/g)')
ax.set_title('Iron Meteorite Classification: Co vs Au')
ax.legend(loc='upper left', ncol=3, fontsize=7, markerscale=0.8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('figures/fig5_co_vs_au.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 分析与讨论

### 1. Co-Au 分布的特征

- **IAB 族**：Au 含量变化范围大（1.3–3.3 μg/g），反映了 IAB 族内部的化学分异。
  - IAB-MG（主群）：中等 Au (~1.5 μg/g)，Co ~4.5–4.7 mg/g
  - IAB-sHL：高 Au (~2.4–3.3 μg/g)，Co ~5.5–5.6 mg/g
  - IAB-sLM：中等 Au (~1.6–1.7 μg/g)，Co ~5.3–5.7 mg/g
- **IIAB**：低 Au (~0.5–1.1 μg/g)，Co ~4.5–5.1 mg/g，形成独立的低Au区域
- **IIIAB**：中等偏低 Au (~0.5–1.1 μg/g)，Co ~5.0–5.2 mg/g
- **IVA**：高 Au (~2.6–2.7 μg/g)，低 Co (~4.3–4.4 mg/g)

### 2. Ga-Au 分布的特征

- **IVA** 类以极低的 Ga (~1.6–1.8 μg/g) 和高 Au (~2.6–2.7 μg/g) 为特征，在图上完全独立
- **IIAB** 具有高 Ga (~52–59 μg/g) 和低到中等 Au (~0.5–1.1 μg/g)，与 IAB 族部分重叠
- **IAB-MG** 具有最高的 Ga (~82–90 μg/g)
- **ungr（未分组）** 类散布在图中的不同区域，部分可能代表新的化学群

### 3. Co、Ga、Au 三元素分类的有效性

Co、Ga 和 Au 三个元素能够有效区分铁陨石化学群，原因在于：

1. **亲铁性差异**：
   - Co 为中等亲铁元素，其分配行为在金属熔体中受温度和氧逸度影响
   - Ga 为中等挥发性亲铁元素，对冷凝/蒸发过程敏感
   - Au 为强亲铁元素，其丰度反映了行星体中金属-硅酸盐分异的程度

2. **地球化学行为差异**：
   - Ga 的挥发性使其成为母体热历史的敏感示踪剂
   - Au 的强亲铁性使其在核幔分异中几乎全部进入金属相
   - Co 兼具亲铁和亲硫特性，对硫化物分异也有响应

3. **母体过程的指纹**：
   - 这三个元素的组合反映了母体形成时的温度、压力条件
   - 不同化学群在这三个元素上的系统差异对应了不同的母体来源
   - Ga/Au 比值是区分 IAB、IIAB 和 IVA 族的极好指标

4. **分类优势**：
   - 相比 Ge-Ni 和 Ir-Ni 图，Co-Au 和 Ga-Au 图能更好地分离某些化学群
   - 特别是 IVA（极低 Ga）、IIAB（高 Ga、低 Au）在 Au-Ga 空间有清晰的分辨率
   - 对数坐标的 Ga-Au 图可有效展示超过两个数量级的 Au 变化
