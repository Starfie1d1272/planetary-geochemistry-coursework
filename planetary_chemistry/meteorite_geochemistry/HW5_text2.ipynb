# HW5 第二题：冰川-海洋氧同位素质量平衡

## 已知条件

- 冰川约占地球表层水体总量的 **2.1%**，δ¹⁸O ≈ **-30‰** (SMOW)
- 剩余 ~97.9% 的表层水体储存于海洋中
- 假设冰川冰的同位素组成保持不变

## 问题 (1)
若冰川总质量增加 50%，海洋水体的同位素组成将会如何变化？

### 质量平衡模型

设：
- $M_{ice}$：当前冰川质量
- $M_{oc}$：当前海洋质量
- $M_{total} = M_{ice} + M_{oc}$：表层水体总质量（守恒）
- $f_{ice} = M_{ice} / M_{total} = 0.021$
- $f_{oc} = M_{oc} / M_{total} = 0.979$

冰川增加50%后：$M_{ice}' = 1.5 \times M_{ice}$，则 $M_{oc}' = M_{total} - M_{ice}'$

氧同位素质量平衡 (当前状态)：
$$\delta_{total} = f_{ice} \cdot \delta_{ice} + f_{oc} \cdot \delta_{oc}$$

冰川增长后的新平衡：
$$\delta_{total} = f_{ice}' \cdot \delta_{ice} + f_{oc}' \cdot \delta_{oc}'$$

由于总质量守恒且冰川同位素组成不变，可以解出新的海洋 δ¹⁸O 值。

```python
# ========== 参数设定 ==========
f_ice_0 = 0.021        # 当前冰川占表层水体比例
f_oc_0 = 1.0 - f_ice_0  # 当前海洋比例 (~0.979)
d18O_ice = -30.0       # 冰川 δ¹⁸O (‰, SMOW)

# 假设当前海洋为 SMOW 组成（δ¹⁸O = 0‰）
d18O_oc_0 = 0.0

# 当前总水体平均同位素组成
d18O_total = f_ice_0 * d18O_ice + f_oc_0 * d18O_oc_0

print(f"当前状态:")
print(f"  冰川比例: {f_ice_0*100:.2f}%")
print(f"  海洋比例: {f_oc_0*100:.2f}%")
print(f"  总水体 δ¹⁸O = {d18O_total:.3f}‰")
print(f"  当前海洋 δ¹⁸O = {d18O_oc_0}‰ (假设为SMOW)")
print(f"  当前冰川 δ¹⁸O = {d18O_ice}‰")
```

```python
# ========== 冰川增加50% ==========
# 新冰川质量 M_ice' = 1.5 * M_ice
# 新总质量 M_total = M_ice' + M_oc'
# 设原总质量为1，则 M_ice_0 = 0.021, M_oc_0 = 0.979
# M_ice_new = 1.5 * 0.021 = 0.0315
# M_oc_new = 1.0 - 0.0315 = 0.9685

f_ice_new = 1.5 * f_ice_0
f_oc_new = 1.0 - f_ice_new

# 从总水体的质量平衡求新海洋 δ¹⁸O
# d18O_total = f_ice_new * d18O_ice + f_oc_new * d18O_oc_new
d18O_oc_new = (d18O_total - f_ice_new * d18O_ice) / f_oc_new
delta_d18O_oc = d18O_oc_new - d18O_oc_0

print(f"冰川增加50%后:")
print(f"  新冰川比例: {f_ice_new*100:.2f}%")
print(f"  新海洋比例: {f_oc_new*100:.2f}%")
print(f"  新海洋 δ¹⁸O = {d18O_oc_new:.4f}‰")
print(f"  海洋 δ¹⁸O 变化 = +{delta_d18O_oc:.4f}‰")
print(f"\n结论: 海洋δ¹⁸O将升高约 {delta_d18O_oc:.2f}‰")
```

```python
# ========== 敏感度分析：不同冰川增长幅度对海洋δ¹⁸O的影响 ==========
import numpy as np

ice_increase = np.linspace(0, 200, 100)  # 冰川增长百分比 0%~200%
f_ice_new_arr = (1 + ice_increase / 100) * f_ice_0
f_oc_new_arr = 1.0 - f_ice_new_arr
d18O_oc_new_arr = (d18O_total - f_ice_new_arr * d18O_ice) / f_oc_new_arr

print("冰川增长幅度与海洋δ¹⁸O的关系:")
print(f"{'冰川增长':>10s}  {'冰川占比':>10s}  {'海洋δ¹⁸O':>12s}")
for inc in [0, 20, 50, 100, 200]:
    idx = np.argmin(np.abs(ice_increase - inc))
    print(f"{inc:>8.0f}%  {f_ice_new_arr[idx]*100:>8.2f}%  {d18O_oc_new_arr[idx]:>10.4f}‰")
```

```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 150

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 子图1: 海洋δ¹⁸O vs 冰川增长
ax1.plot(ice_increase, d18O_oc_new_arr, 'b-', linewidth=2)
ax1.axvline(x=50, color='red', linestyle='--', alpha=0.7, label='+50% (this scenario)')
ax1.axhline(y=0, color='gray', linestyle=':', alpha=0.5)
ax1.set_xlabel('Glacier mass increase (%)')
ax1.set_ylabel('Seawater $\delta^{18}$O (‰, SMOW)')
ax1.set_title('Seawater $\delta^{18}$O vs Glacier Growth')
ax1.legend()
ax1.grid(True, alpha=0.3)

# 子图2: 质量平衡示意图
ax2.barh(['Initial ocean', 'Initial ice'], [f_oc_0 * 100, f_ice_0 * 100],
         color=['#2196F3', '#B3E5FC'], edgecolor='black', height=0.5)
ax2.barh(['New ocean', 'New ice'], [f_oc_new * 100, f_ice_new * 100],
         color=['#1565C0', '#81D4FA'], edgecolor='black', height=0.5)
ax2.set_xlabel('Fraction of total surface water (%)')
ax2.set_title('Mass redistribution (Ice +50%)')

plt.tight_layout()
plt.savefig('figures/fig3_glacier_sensitivity.png', dpi=200, bbox_inches='tight')
plt.show()
```

## 问题 (2)

### 文献调研：地球历史上海水氧同位素的变化

**核心问题**：如果海水的 δ¹⁸O 曾经为 **+3‰**，能否通过扩大冰川量实现？

#### 理论计算

从质量平衡方程反算：要使得海水 δ¹⁸O = +3‰，需要多少冰川？

$$\delta_{total} = f_{ice} \cdot (-30) + (1-f_{ice}) \cdot 3$$
$$f_{ice} = \frac{3 - \delta_{total}}{3 + 30}$$

其中 $\delta_{total}$ 为总水体平均同位素组成（在当前状态下计算）。

```python
# ========== 反算：海洋 δ¹⁸O = +3‰ 需要多少冰川? ==========

d18O_oc_target = 3.0  # 目标海水 δ¹⁸O (‰)

# 从质量平衡反算 f_ice:
# d18O_total = f_ice * d18O_ice + (1-f_ice) * d18O_oc_target
# d18O_total = f_ice * (-30) + (1-f_ice) * 3
# d18O_total = -30*f_ice + 3 - 3*f_ice = 3 - 33*f_ice
# f_ice = (3 - d18O_total) / 33

f_ice_for_3permil = (d18O_oc_target - d18O_total) / (d18O_oc_target - d18O_ice)
f_oc_for_3permil = 1.0 - f_ice_for_3permil

print(f"要使海水 δ¹⁸O = +3‰:")
print(f"  需要冰川占比: {f_ice_for_3permil*100:.2f}%")
print(f"  对应海洋占比: {f_oc_for_3permil*100:.2f}%")
print(f"  冰川相对当前增加: {(f_ice_for_3permil/f_ice_0 - 1)*100:.1f}%")

# 海平面下降估算 (当前冰川若全部融化 ≈ 65m 海平面上升)
# 冰川质量 ∝ 海平面当量
sea_level_equivalent_current = 65  # m (当前冰川全部融化造成的海平面上升)
ice_mass_current = f_ice_0
ice_mass_needed = f_ice_for_3permil
additional_ice = ice_mass_needed - ice_mass_current
sea_level_drop = additional_ice / ice_mass_current * sea_level_equivalent_current

print(f"\n若当前冰川全部融化 ≈ {sea_level_equivalent_current} m 海平面上升:")
print(f"  额外需要的冰川 ≈ 当前冰川的 {additional_ice/ice_mass_current*100:.0f}%")
print(f"  相当海平面下降约 {sea_level_drop:.0f} m")
```

### 文献调研：地质历史时期海水 δ¹⁸O 的可能变化

#### 关键科学问题

海水 δ¹⁸O 在地质历史时期的变化受两个主要因素控制：
1. **冰量效应 (Ice Volume Effect)**：大陆冰盖的增长使海水 δ¹⁸O 升高
2. **温度效应**：海水温度变化引起的分馏
3. **水-岩交换**：洋中脊热液系统与洋壳的水岩反应

#### 重要文献与数据

1. **更新世冰期-间冰期旋回**：海水 δ¹⁸O 波动约 **1.0-1.2‰**（Lear et al., 2000; Waelbroeck et al., 2002）
   - 末次盛冰期(LGM): 海水 δ¹⁸O ≈ +1.0‰
   - 间冰期: 海水 δ¹⁸O ≈ 0‰

2. **新生代长期演化**（Zachos et al., 2001, Science）：
   - 早始新世(Early Eocene, ~50 Ma): 底栖有孔虫 δ¹⁸O ≈ 0‰（无大冰盖）
   - 渐新世(Oligocene, ~34 Ma): 南极冰盖开始形成，δ¹⁸O 升高 ~1‰
   - 中新世(Miocene, ~15 Ma): δ¹⁸O 进一步升高
   - 更新世(Pleistocene): 北半球冰盖形成，δ¹⁸O 升高至 ~3‰（但含温度信号）

3. **前寒武纪海水 δ¹⁸O 争议**：
   - 传统观点：前寒武纪海水 δ¹⁸O ≈ 0‰（如现代）
   - 另类观点（Knauth & Lowe, 2003; Jaffrés et al., 2007）：早期海洋 δ¹⁸O 可能更低（-10‰ 至 -3‰），或更高
   - 争论核心：古老燧石的 δ¹⁸O 记录是海水组成信号还是成岩改造信号

#### 海水 δ¹⁸O = +3‰ 是否可达？

从理论计算来看：
- 若 δ¹⁸O_ice = -30‰，要达到海水 δ¹⁸O = +3‰，需要冰川占比约 **9.5%**
- 这相当于当前冰川量的 **~4.5倍**
- 对应的海平面下降将超过 **200 m**

**地质约束**：
- 末次盛冰期(LGM)海平面下降约 120-130 m，海水 δ¹⁸O ≈ +1.0‰
- "雪球地球"(Snowball Earth)事件期间（~720-635 Ma），海平面可能下降更多，但不确定
- **单纯通过冰川扩张使海水 δ¹⁸O 达到 +3‰ 在物理上困难**：需要远超任何已知冰期的冰川规模
- 其他因素（如洋中脊热液蚀变、水岩交换速率变化）可能在更长的时间尺度上改变海水 δ¹⁸O

**结论**：单纯通过冰川扩张将海水 δ¹⁸O 推高至 +3‰ 不太现实，但若考虑地质时间尺度上水-岩交换的累积效应，海水 δ¹⁸O 可能出现偏离 0‰ 的趋势。这是一个仍在争议中的科学问题。

```python
# ========== 示意图：冰量与海水δ¹⁸O关系 ==========
fig, ax1 = plt.subplots(1, 1, figsize=(9, 6))

# 冰量-海水δ¹⁸O关系曲线
f_ice_range = np.linspace(0.01, 0.20, 200)
d18O_oc_range = (d18O_total - f_ice_range * d18O_ice) / (1.0 - f_ice_range)

ax1.plot(f_ice_range * 100, d18O_oc_range, 'b-', linewidth=2, label='Mass balance model')

# 标注关键点
# 当前
ax1.plot(f_ice_0 * 100, d18O_oc_0, 'go', markersize=10, label=f'Present ({f_ice_0*100:.1f}% ice, $\delta^{18}$O={d18O_oc_0}‰)')
# +50% 情景
ax1.plot(f_ice_new * 100, d18O_oc_new, 's', color='purple', markersize=9,
         label=f'Ice +50% ({f_ice_new*100:.2f}% ice, $\delta^{18}$O≈{d18O_oc_new:.2f}‰)')
# LGM
f_ice_LGM = f_ice_0 + 120/65 * f_ice_0
d18O_oc_LGM = (d18O_total - f_ice_LGM * d18O_ice) / (1.0 - f_ice_LGM)
ax1.plot(f_ice_LGM * 100, d18O_oc_LGM, 's', color='orange', markersize=9,
         label=f'LGM (~{f_ice_LGM*100:.1f}% ice, $\delta^{18}$O≈{d18O_oc_LGM:.2f}‰)')
# +3‰ 情景
ax1.plot(f_ice_for_3permil * 100, 3.0, 'r*', markersize=14, markeredgewidth=1.5,
         markeredgecolor='darkred',
         label=f'$\delta^{18}$O=+3‰ needs {f_ice_for_3permil*100:.1f}% ice')

ax1.axhline(y=0, color='gray', linestyle=':', alpha=0.5)
ax1.set_xlabel('Ice fraction of total surface water (%)')
ax1.set_ylabel('Seawater $\delta^{18}$O (‰, SMOW)')
ax1.set_title('Ice Volume vs Seawater $\delta^{18}$O')
ax1.legend(loc='lower right')
ax1.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('figures/fig4_ice_volume_seawater.png', dpi=200, bbox_inches='tight')
plt.show()
```
